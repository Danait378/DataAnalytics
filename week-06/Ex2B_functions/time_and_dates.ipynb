# Exercise 2.B Working with datetime functions
## Lab 2. Times and Dates
This lab practices using Python datetime functions.

```python
import datetime
```

```python
# Today
```

```python
today = datetime.datetime.now()

print(today)
```

```python
formatted_today = today.strftime("%A, %B %d, %Y")

print("Today is", formatted_today)
```

```python
# My birthday
```

```python
birthday = datetime.date(1998,11,20)

print(birthday)
```

```python
birthday_format = birthday.strftime("%m/%d/%Y")

print("My birthday is", birthday_format)
```

```python
# 90 days
```

```python
ninety_d = today + datetime.timedelta(days=90)

print(ninety_d)
```

```python
future_date = ninety_d.strftime("%B %d, %Y")

print("90 days from today is", future_date)
```

```python
# Time for Dinner
```

```python
dinner_time = datetime.time(6, 30)

formatted_dinner = dinner_time.strftime("%I:%M %p")

print("Let's meet for dinner at", formatted_dinner)
```
