# Exercise 2.C Defining Custom Functions
## Lab 1 - Simple Functions

This lab is about creating simple custom functions in Python and practicing how to use parameters.

```python
def favorite_things(name, movie, music):

    print("My name is", name)
    print("I really like the movie", movie)
    print("My favorite artist or band is", music)


favorite_things("Danait", "prison Break", "Mark")
```

```python
def why_im_here(purpose):

    print(purpose)


why_im_here("I joined Year Up United to learn data analytics skills and prepare for my future career.")
```

```python
def favorite_place(place, why):

    return f"One of my favorite places is (place) because (why)."
```

```python
print(favorite_place("Hilton Island", "there are many fun things to do"))
```

```python
print(favorite_place("the beach", "it helps me relax and enjoy the view"))
```
