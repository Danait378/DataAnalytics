# Lab 3 - Tax Functions

# This lab shows how to use functions to calculate
# Social Security tax, Medicare tax, and Federal tax.

```python
# Function for Social Security tax

def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
# Medicare tax function

def get_medicare_tax(pay):
    tax = pay * 0.0145
    return tax
```

```python
# Federal tax function

def get_federal_tax(pay, code):

    if code == 0:
        rate = 0.23

    elif code == 1:
        rate = 0.21

    elif code == 2:
        rate = 0.195

    elif code == 3:
        rate = 0.185

    else:
        rate = 0.18 - ((code - 4) * 0.005)

    tax = pay * rate 
    return tax
    
```

```python
# Person 1

pay = 750
code = 0

print("Person 1")
print("Social Security Tax:", round(get_soc_sec_tax(pay), 2))
print("Medicare Tax:", round(get_medicare_tax(pay), 2))
print("Federal Tax:", round(get_federal_tax(pay, code), 2))
```
