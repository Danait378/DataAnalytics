### Exercise 2.E Using Functions for Analyzing Data

This notebook looks at a small sales dataset using Python functions.
The data includes product names, sales regions, and sale amounts.


```python
sales_data = [
    ('Laptop', 'East', 1200.00),
    ('Phone', 'West', 850.00),
    ('Tablet', 'East', 400.00),
    ('Monitor', 'South', 650.00),
    ('Keyboard', 'West', 75.00),
    ('Laptop', 'South', 1150.00),
    ('Phone', 'East', 900.00),
    ('Tablet', 'West', 425.00),
    ('Monitor', 'East', 680.00),
    ('Keyboard', 'South', 80.00),
    ('Laptop', 'West', 1300.00),
    ('Phone', 'West', 820.00)
]

print(sales_data)
```

## Lab 1. Total sales by region

```python
def region_total(data, region):

    total = 0

    for product, sales_region, amount in data:

        if sales_region == region:
            total += amount
            return total


```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

```python
## Lab 2. Count of sales by region
```

```python
def region_count(data, region):

    count = 0

    for product, sales_region, amount in data:

        if sales_region == region:
            count += 1

    return count
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):

    total = region_total(data, region)
    count = region_count(data, region)

    if count == 0:
        return 0

    return total / count
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'South'))
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):

    results = []

    for record in data:

        if record[2] > threshold:
            results.append(record)

    return results
```

```python
high_sales = sales_above(sales_data, 500)

for record in high_sales:
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    return max(data, key=lambda s: s[2])


def bottom_sale(data):
    return min(data, key=lambda s: s[2])
```

```python
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

The highest sale was a Laptop in  West region for $1300.00.

The lowest sale was a Keyboard in West region for $75.00.

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):

    updated_sales = []

    for product, region, amount in data:

        new_amount = amount - (amount * discount_pct)

        updated_sales.append((product, region, new_amount))

    return updated_sales
```

```python
discounted_sales = apply_discount(sales_data, 0.15)

print("Original:")
for sale in sales_data[:3]:
    print(sale)

print()

print("Discounted:")
for sale in discounted_sales[:3]:
    print(sale)
```

## Lab 7. Unique regions

```python
def unique_regions(data):

    all_regions = []

    for product, region, amount in data:

        if region not in all_regions:
            all_regions.append(region)

    return sorted(all_regions)
```

```python
print(unique_regions(sales_data))
```

This function can help organize and analyze sales data by region.
It could also be useful for creating reports, charts, or checking which regions are included in the dataset.

## Lab 8. Region summary

```python
def region_summary(data, region):

    region_data = []

    for sale in data:

        if sale[1] == region:
            region_data(sale)

    top = top_sale(region_data)

    print(region, "Region")
    print("Number of sales:", region_count(data, region))
    print("Total sales: $", format(region_total(data, region), ".2f"))
    print("Average sale: $", format(region_average(data, region), ".2f"))
    print("Top sale:", top[0], "- $", format(top[2], ".2f"))
```

```python
region_summary(sales_data, 'East')
```

## Lab 9. Full summary

```python
for region in unique_regions(sales_data):
    region_summary(sales_data, region)
    print()
```
