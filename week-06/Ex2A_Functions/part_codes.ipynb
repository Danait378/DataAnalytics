# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):

    supplier = part_code.split(":")[0]

    return supplier
```

```python
def get_product_num(part_code):

    product = part_code.split(":")[1]
    product_num = product.split("-")[0]

    return product_num
```

```python
def get_size(part_code):

    size = part_code.split("-")[1]

    return size
```

```python
part1 = "ACME:123-L"
part2 = "DI:12345-M"
part3 = "ACE:1-12"
```

```python
supplier = get_supplier_code(part1)
product = get_product_num(part1)
size = get_size(part1)

print("For", part1 + ",", "the supplier is", supplier + ".", 
      "Part number", product, "is available in size", size + ".")
```

```python
supplier = get_supplier_code(part2)
product = get_product_num(part2)
size = get_size(part2)

print("For", part2 + ",", "the supplier is", supplier + ".", 
      "Part number", product, "is available in size", size + ".")
```
