```python
import numpy as np
```

## Sample Arrays

```python
# Random arrays to practice with

r_1 = np.random.randint(1, 21, 2)

r_2 = np.random.randint(1, 21, (1, 2))

r_3 = np.random.random((2, 1))

r_4 = np.random.randint(20, 101, (4, 5))

r_5 = np.random.random((4, 5))

print(r_1)
print(r_2)
print(r_3)
print(r_4)
print(r_5)
```

## Permutations

```python
# Print r_4 before shuffling
print(r_4)
```

```python
# Shuffle the rows in r_4
np.random.shuffle(r_4)

print(r_4)
```

The shuffle function changes the order of the rows in the array.  
It also changes the original array itself.

```python
# Create a shuffled copy
new_shuffle = np.random.permutation(r_4)

print(new_shuffle)

# Print r_4 again
print(r_4)
```

Shuffle changes the original array.  
Permutation makes a new shuffled copy and keeps the original array the same.

## Mathematical Operations

```python
print(r_1)
print(r_2)

print(r_1.shape)
print(r_2.shape)

answer_1 = np.add(r_1, r_2)

print(answer_1)
print(answer_1.shape)
```

```python
print(r_1)
print(r_3)

print(r_1.shape)
print(r_3.shape)

answer_2 = np.add(r_1, r_3)

print(answer_2)
print(answer_2.shape)
```

```python
print(r_2)
print(r_4)

print(r_2.shape)
print(r_4.shape)

answer_3 = np.add(r_2, r_4)

print(answer_3)
print(answer_3.shape)
```

```python
# This gives an error because the array sizes do not match

print(np.add(r_1, r_4))
```

```python
# Add [[1]] to every value in r_4
print(np.add(r_4, np.array([[1]])))
```

```python
# Add 1 to each column
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

Broadcasting is how NumPy works with arrays of different sizes.
NumPy matches the shapes so math operations can work together.
The larger array usually decides the shape of the final result.

## Cumulative Operations

```python
print(r_4)
```

```python
# Add all values
print(np.sum(r_4))
```

```python
# Add values down the columns
print(np.sum(r_4, axis=0))
```

```python
# Add values across the rows
print(np.sum(r_4, axis=1))
```

```python
# Rolling total
print(np.cumsum(r_4))
```

```python
# Rolling total down the columns
print(np.cumsum(r_4, axis=0))
```

## Grocery Problem

```python
# vegetables, fruits, snacks, beverages

bag1 = np.array([2, 3, 4, 0])

bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
# Total items in each bag

total1 = np.sum(bag1)
total2 = np.sum(bag2)

print("The first bag contains", total1, "items.")
print("The second bag contains", total2, "items.")
```

```python
# Total items in each category

totals = bag1 + bag2

print("There are", totals[0], "vegetables,", 
      totals[1], "fruits,", 
      totals[2], "boxes of snacks, and", 
      totals[3], "beverages.")
```

## Teamwork

```python
# Ali's data
ali = np.array([5, 4, 6, 4, 2])

print(ali)
```

```python
# Ben's data
ben = np.array([0, 5, 5, 3, 4])

print(ben)
```

```python
# Carla's data
carla = np.array([3, 6, 0, 3, 4])

print(carla)
```

```python
# Put all data together

all_knocks = np.array([ali, ben, carla])

print(all_knocks)
```

```python
# Weekly totals

print("Ali knocked on", np.sum(ali), "doors.")
print("Ben knocked on", np.sum(ben), "doors.")
print("Carla knocked on", np.sum(carla), "doors.")

print("The group knocked on", np.sum(all_knocks), "doors total.")
```

```python
# Daily totals
daily_total = np.sum(all_knocks, axis=0)

# Rolling total
rolling_total = np.cumsum(daily_total)

# Estimated votes
estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
