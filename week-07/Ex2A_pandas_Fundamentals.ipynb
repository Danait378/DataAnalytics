# Exercise 2.A Fundamentals of pandas

```python
import numpy as np
import pandas as pd
```

## Series

```python
# Different Python data types

aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    "first": 13,
    "second": 14,
    "third": 15,
    "fourth": 16
}
```

```python
# Turn the Python objects into pandas Series

ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac))
ser4 = pd.Series(ad)

print(ser1)
print()

print(ser2)
print()

print(ser3)
print()

print(ser4)
```

```python
# This gives an error because the values are not written correctly

ser5 = pd.Series(17, 18, 19, 20)

# TypeError: Series.__init__() takes from 1 to 4 positional arguments but 5 were given
```

## DataFrames from Series

```python
# Create a DataFrame from ser1

df1 = pd.DataFrame(ser1)

df1
```

```python
# Add new columns using insert()

df1.insert(1, "Second", ser2)

df1.insert(2, "Third", ser3)

df1
```

```python
# Add another column using ser4

df1["Fourth"] = ser4

df1

# ser4 uses words as indexes like first and second.
# The other Series use number indexes like 0,1,2,3.
# Because the indexes do not match, pandas shows NaN values.
```

```python
# Add a new row using loc

df1.loc[4] = [5, 9, 12, 0]

df1
```

## Coding a DataFrame

```python
# Create dictionary data

test_data = {
    "ID": [1, 2, 3, 4, 5, 6, 7, 8],

    "Name": ["Alicia", "Bob", "Charlie", "Dominic",
             "Eve", "Frank", "Grace", "Heidi"],

    "Age": [25, np.nan, 22, 19, 31, 35, np.nan, 18],

    "City": ["NY", "LA", "Chicago", "Houston",
             "Phoenix", "Boston", "Austin", "San Diego"],

    "Score": [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Create DataFrame

df4 = pd.DataFrame(test_data)

df4
```

```python
# Export to CSV file

df4.to_csv("test_data.csv", index=False)
```
