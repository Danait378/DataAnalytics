```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row of the array and multiplies the whole row by 2
The second loop goes inside each row and multiplies each individual number by 2

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This prints the whole row doubled each time because loop_result uses x * 2, not y * 2.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA,), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
   
Error all input arrays must have same shape
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
h_stack = np.hstack((arrA, arrA))  # joins arrays horizontally
v_stack = np.vstack((arrA, arrA))  # joins arrays vertically
d_stack = np.dstack((arrA, arrA))  # joins arrays by depth

print(h_stack)
print(v_stack)
print(d_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# This breaks the array into 2 parts. Each part still has rows and columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# This breaks the array into 3 smaller parts. Each part still has rows and columns, but some parts may be bigger than others.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])

arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7
arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7
arrD = arrB[y]

print(arrD)
```
