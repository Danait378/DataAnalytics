```python
import numpy as np
```

## Generating Random Numbers

```python
# Random integers from 1 to 20
r_1 = np.random.randint(1, 21, 2)

print(r_1)
```

```python
# Random integers with 1 row and 2 columns
r_2 = np.random.randint(1, 21, (1, 2))

print(r_2)
```

```python
# Random decimal numbers between 0 and 1
r_3 = np.random.random((2, 1))

print(r_3)
```

```python
# Random integers from 20 to 100
r_4 = np.random.randint(20, 101, (4, 5))

print(r_4)
```

```python
# Random decimal numbers between 0 and 1
r_5 = np.random.random((4, 5))

print(r_5)
```

## Generating Random Choices from an Array

```python
# List of student names
students = np.array([
    "Alexus", "Blake,", "Corey", "Daniel",
    "Lorah", "Hana", "Louis", "mohamed",
    "Saliem", "Tihtina", "samia", "victoria"
])

print(students)
```

```python
# Pick one random student
call_on = np.random.choice(students)

print(call_on)
```

```python
# Create random breakout groups
breakout_groups = np.random.choice(students, (3, 4))

print(breakout_groups)

# This may not be  best way to make groups
# because some students can appear more than once.
```

```python
# Pick 5 students with different chances
call_on_weighted = np.random.choice(
    students,
    5,
    p=[0.30, 0.10, 0.10, 0.10,
       0.05, 0.05, 0.05, 0.05,
       0.05, 0.05, 0.10, 0.00]
)

print(call_on_weighted)

# Alexus has a 30% chance of being picked.
# Victodia has a 0% chance, so they will not be picked.
# Blake has 10% chance of being picked.
# The other students have smaller chances.
```

## Generating a Normal Distribution

```python
# Create a normal distribution
bell_curve = np.random.normal(10, 5, (4, 5))

print(bell_curve)
```
