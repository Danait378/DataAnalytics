working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

df_sales = pd.DataFrame(data)
df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
total_units = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()

print("Total units sold:", total_units)
print("Total revenue:", total_revenue)
```

```python
average_units = df_sales['Units_Sold'].mean()
average_revenue = df_sales['Revenue'].mean()

print("Average units sold:", average_units)
print("Average revenue:", average_revenue)
```

## Grouping Data

```python
sales_by_product = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
sales_by_product
```

```python
average_by_region = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
average_by_region
```

## Sorting and Ranking Data

```python
df_sales.sort_values('Units_Sold', ascending=True)
```

```python
df_sales.sort_values('Revenue', ascending=False)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)
df_sales
```

## Merging DataFrames

```python
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}

df_products = pd.DataFrame(product_data)
df_products
```

```python
df_merged = pd.merge(df_sales, df_products, on='Product')
df_merged
```

```python
revenue_by_category = df_merged.groupby('Category')['Revenue'].sum()
revenue_by_category
```
