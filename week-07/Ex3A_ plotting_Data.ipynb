# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
# Data points
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
# Basic line graph
plt.plot(x, y)
plt.show()
```

```python
# Line graph with markers
plt.plot(x, y, marker='o')
plt.show()
```

```python
# Customized graph
plt.plot(x, y, marker='o')

plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
# Dashed red line
plt.plot(x, y, linestyle='--', color='red')
plt.show()
```

```python
# Green square markers
plt.plot(x, y, marker='s', markersize=10, markerfacecolor='green')
plt.show()
```

```python
# Different marker style and color
plt.plot(x, y, marker='^', color='blue')
plt.show()

# Another marker style and color
plt.plot(x, y, marker='*', color='purple')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# Product names and sales numbers
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

# Create a simple bar chart
plt.bar(products, sales)

# Show the chart
plt.show()
```

```python
# Create a customized bar chart
plt.bar(products, sales, color='skyblue', width=0.5)

# Add a title and labels
plt.title("Sales for Each Product")
plt.xlabel("Products")
plt.ylabel("Number of Sales")

# Show the chart
plt.show()
```

```python
# Create random x and y values
x = np.random.rand(50)
y = np.random.rand(50)

# Create a basic scatter plot
plt.scatter(x, y)

# Show the plot
plt.show()
```

```python
# Create random data for the scatter plot
x = np.random.rand(50)
y = np.random.rand(50)

# Random sizes for the dots
sizes = np.random.rand(50) * 100

# Random colors for the dots
colors = np.random.rand(50)

# Create the scatter plot
plt.scatter(x, y, s=sizes, c=colors, cmap='viridis')

# Add labels and title
plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")

# Show the plot
plt.show()
```

## Subplots

```python
# Create data for a sine wave
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create random numbers for the histogram
random_numbers = np.random.randn(1000)

# Create a figure with 2 plots side by side
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# First subplot: line graph
ax[0].plot(x, y)
ax[0].set_title("Sine Wave")
ax[0].set_xlabel("X Values")
ax[0].set_ylabel("Y Values")

# Second subplot: histogram
ax[1].hist(random_numbers)
ax[1].set_title("Histogram Example")
ax[1].set_xlabel("Numbers")
ax[1].set_ylabel("Frequency")

# Display both plots
plt.show()
```
